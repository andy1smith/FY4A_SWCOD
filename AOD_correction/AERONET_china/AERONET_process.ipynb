```python
import pandas as pd
import numpy as np

# 1. Load the AERONET data, skipping the 6 lines of metadata
site='BJC_CAMS'
name = 'Beijing-CAMS'

file_path = f'20210101_20211231_{name}.lev20'
df = pd.read_csv(file_path, skiprows=6)

# 2. Combine the Date and Time columns into a single datetime object
# AERONET format is dd:mm:yyyy and hh:mm:ss
datetime_str = df['Date(dd:mm:yyyy)'] + ' ' + df['Time(hh:mm:ss)']
parsed_datetime = pd.to_datetime(datetime_str, format='%d:%m:%Y %H:%M:%S')

# 3. Format the time to your requested string format (YYYY-MM-DD HH:MM:SS.000)
# Note: AERONET data only provides seconds, so the milliseconds will be padded with zeros.
df['time'] = parsed_datetime.dt.strftime('%Y-%m-%d %H:%M:%S.000')

# 4. Extract only the 'time' and 'AOD_500nm' columns
df_extracted = df[['time', 'AOD_500nm']].copy()

# 5. Handle missing values 
# AERONET uses -999.0 for missing data. It's best practice to convert these to NaN.
df_extracted['AOD_500nm'] = df_extracted['AOD_500nm'].replace(-999.0, np.nan)

# Display the first few rows to verify
# Define your requested filename
output_filename = f'2021_{site}.csv'

# Save the two-column dataframe to CSV
df_extracted.to_csv(output_filename, index=False)

print(f"Successfully saved {len(df_extracted)} rows to {output_filename}")
```
